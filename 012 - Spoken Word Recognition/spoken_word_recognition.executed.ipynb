## **1 - Business Objective**

### **1.1 - Overview**

A voice interface on a phone, a car dashboard, or a home speaker has to decide, from about one second
of audio, which of a small fixed vocabulary a person just said. The vocabulary is deliberately small,
in the order of tens of words, because the decision runs on the device itself, wakes the rest of the
system, and has to answer before the person notices a delay. A cloud round trip is unavailable when
the network is down and undesirable when the audio is private.

The engineering question is what representation of the audio the on-device classifier should consume.
Raw audio at 16 kHz gives 16,000 amplitudes per second per clip, which is expensive to store and
expensive to compute over. The standard compression is a spectrogram, and the standard spectrogram for
speech is the Mel-Frequency Cepstral Coefficient representation. That still leaves a choice the
engineer makes early and rarely revisits: keep the full time-by-frequency grid, or average it down the
time axis into one compact vector per clip. The first choice costs 32 times the memory and 32 times
the input width. The second is cheaper on every axis that matters on a device.

This notebook makes that choice an experiment rather than an assumption, and then asks what the model's
own explanations say about why the answer came out the way it did.

### **1.2 - Business Objective Statement**

Given one second of 16 kHz audio containing a single spoken word drawn from a fixed twenty-word
vocabulary (the digits zero through nine and the command words yes, no, up, down, left, right, on, off,
stop, and go), classify the word, and determine which of two MFCC feature representations, a
time-averaged forty-coefficient vector or the full forty-by-thirty-two spectrogram, a production
on-device classifier should use.

## **2 - Problem Statement**

### **2.1 - Overview**

This is a supervised multiclass classification problem with twenty balanced classes. The input is one
second of mono audio sampled at 16 kHz. The target is the word identity. The dataset is Google Speech
Commands v0.02, released by Pete Warden in 2018, which contains roughly 106,000 one-second clips of 35
words recorded by volunteers through browsers and phones in uncontrolled acoustic conditions. The
twenty-word subset used here holds roughly 77,000 clips contributed by roughly 2,580 speakers.

Two properties of the data drive the experimental design.

First, every clip carries the recording speaker's identity in its filename, as an eight-character hash.
A random clip-level split puts the same speaker's voice on both sides of the train and test boundary,
and a model that memorizes a speaker's vocal tract then scores well without having learned anything
about words. Section 8.6 splits on the speaker hash instead, so no speaker appears in more than one
partition.

Second, the recordings are uncontrolled. Clips vary in loudness, in background noise, in microphone,
and in where within the one-second window the word actually falls. That variability is the reason the
representation question has a non-obvious answer, and it is why a result measured on this dataset says
something about deployment rather than about a laboratory.

A secondary label, the recording speaker's hash, is used only to construct the split. It never enters
the training features and never enters the target.

## **3 - Solution Methodology**

### **3.1 - Overview**

The notebook proceeds in stages. First, the Speech Commands archive is downloaded, indexed by word and
by speaker hash, and reduced to the twenty-word subset. Second, MFCCs are extracted at forty mel
coefficients across thirty-two time frames, and each clip is turned into two parallel feature vectors,
a mean-pooled vector of forty numbers and a flattened vector of 1,280 numbers. Third, human-engineered
and AI-suggested candidate features are layered on top, with an ablation measuring what each source
contributes. Fourth, four models are trained under both representations and the same speaker-disjoint
protocol: a Decision Tree, a Random Forest, XGBoost, and a light PyTorch feedforward network, with a
pretrained wav2vec 2.0 benchmark alongside them. Fifth, permutation importance is computed over the
spectrogram grid and aggregated separately down the mel axis and along the time axis. Sixth, a
LangGraph agent turns a low-confidence prediction into a plain-language recommendation for the engineer
who owns the wake-word pipeline, and an autonomous variant of that agent adds planning, tool use, and
memory.

The four-model progression matters for a reason beyond benchmarking. A neural network is the expensive
choice for an on-device classifier, and it earns its place only by beating a properly tuned gradient
boosting model by a margin large enough to justify the cost. The comparison is run so that the reader
derives that verdict rather than accepts it.

### **3.2 - The Rediscovery Standard**

The model is never told that the time-course of the spectrum carries word identity. It is given MFCC
numbers and a word label, nothing more. Nothing in the pipeline encodes that the transition between
one speech sound and the next is where a listener finds the consonant, or that averaging a spectrogram
down its time axis destroys exactly that information while leaving the average spectral envelope
intact.

Section 15 checks whether the model's own behavior recovers the finding. Two things have to hold. The
full spectrogram representation has to beat the mean-pooled representation by a margin larger than the
extra parameters alone would explain, and permutation importance over the grid has to concentrate on
particular time frames rather than spreading evenly across them. If both hold, the model has
independently recovered what Delattre, Liberman, and Cooper established at Haskins Laboratories in
1955 by playing hand-painted spectrograms back as sound.

If either fails, this section reports the failure and the law is dropped from the case study. CS4 set
that precedent when Granovetter's Strength of Weak Ties was tested, failed to hold on the data, and was
removed rather than softened.

## **4 - A Brief History of Automatic Speech Recognition**

### **4.1 - Overview**

Machine speech recognition began as an engineering problem about signals and became a problem about
representation. The instruments arrived first. Homer Dudley demonstrated the Voder, a hand-operated
speech synthesizer, at the 1939 New York World's Fair, and his colleagues at Bell Telephone
Laboratories followed it with the sound spectrograph, which drew speech as a picture of frequency
against time. Ralph Potter, George Kopp, and Harriet Green published that work as Visible Speech in
1947, and for the first time a researcher could look at a word.

What they saw was motion. A vowel held steady produced horizontal bands, the formants, at frequencies
set by the shape of the vocal tract. A consonant produced a bend in those bands as the tract moved into
or out of the vowel. The bend was brief and it was where the consonant lived.

Five years later, K. H. Davis, R. Biddulph, and Stephen Balashek built Audrey at Bell Labs and taught
it the ten spoken digits by tracking exactly those formants. Audrey worked reliably for the one speaker
it was tuned to. Retuned for a second speaker it degraded, and for an untrained speaker it failed. That
raised the question the next three decades answered: if the machine was already measuring the right
quantity, why did the answer not survive a change of voice?

Haskins Laboratories supplied the answer in 1955, and it was not about better measurement. Pierre
Delattre, Alvin Liberman, and Franklin Cooper found that no fixed spectral pattern corresponds to a
consonant at all. The same steady spectrum produces a different consonant depending on how the formants
move into the vowel that follows it. Audrey had been reading the wrong axis. What identified a word
was not the picture but the change in the picture over time, and a system that averages the picture
never sees it.

That is the finding this notebook hands to a model with the label removed. The model receives MFCC
numbers and word labels. It is never told that the time axis carries the word. Section 15 asks whether
it works that out anyway.

### **4.2 - Milestones in Automatic Speech Recognition**

The milestones below trace one continuous argument. The instruments come first, the mistaken
representation follows, Haskins corrects it in 1955, and every architecture after that is a
progressively better bet on the time axis.

| Year | Who | What |
|---|---|---|
| 1937 | Stanley Smith Stevens, John Volkmann, and Edwin Newman | Publish the mel scale in the Journal of the Acoustical Society of America, mapping measured frequency to perceived pitch and establishing that human frequency resolution is roughly logarithmic above about 500 Hz |

| 1947 | Ralph Potter, George Kopp, and Harriet Green, Bell Labs | Publish Visible Speech, introducing the sound spectrograph and the practice of reading speech from a time-frequency picture |
| 1952 | Stephen Balashek, R. Biddulph, and K. H. Davis, Bell Labs | Build Audrey, which recognizes the spoken digits zero through nine from a single speaker by tracking formant frequencies |
| 1955 | Pierre Delattre, Alvin Liberman, and Franklin Cooper, Haskins Laboratories | Publish the acoustic-loci and formant-transition work, showing with hand-painted spectrograms played back through the Pattern Playback that the same steady spectrum yields different consonants depending on the transition into the vowel |

Read down the right-hand column and the same correction repeats at increasing cost. Dynamic time
warping aligned the time axis by hand. Hidden Markov models modelled it as a state sequence after
Baker and Jelinek. Recurrent networks and then transformers modelled it with learned parameters. Every
one of those steps buys the same thing Haskins identified with painted acetate in 1955, and every one
of them is expensive. An engineer who averages a spectrogram down its time axis to save memory is
undoing all of it in one line.

## **5 - The Science: The Mel Scale, MFCCs, and Spectro-Temporal Structure**

### **5.1 - The Mel Scale**

Stevens, Volkmann, and Newman asked listeners in 1937 to adjust a tone until it sounded twice the pitch
of a reference, and found that perceived pitch grows more slowly than measured frequency. The scale
they fitted to those judgments is the mel scale, conventionally written as

$$ m = 2595 \, \log_{10}\!\left(1 + \frac{f}{700}\right) $$

where $f$ is frequency in hertz and $m$ is pitch in mels. Below roughly 500 Hz the mapping is close to
linear. Above it, equal steps in mels correspond to increasingly wide steps in hertz.

The practical consequence is that a filterbank spaced evenly in mels puts narrow filters where speech
carries the information a listener uses, in the first two or three kilohertz where the first and second
formants sit, and wide filters where it does not. A forty-filter mel bank is therefore a far better use
of forty numbers than forty evenly spaced linear filters would be.

### **5.2 - From Waveform to MFCC**

The Davis and Mermelstein pipeline of 1980 turns a waveform into a compact matrix in six steps.

1. Pre-emphasis. A first-order high-pass filter lifts the higher frequencies, compensating for the
   roughly 6 dB per octave roll-off of the glottal source.
2. Framing. The signal is cut into short overlapping frames, typically 25 ms long with a 10 ms hop, so
   that the spectrum within a frame can be treated as stationary.
3. Windowing. Each frame is multiplied by a Hann or Hamming window to suppress the spectral leakage a
   hard frame boundary would produce.
4. Magnitude spectrum. A discrete Fourier transform per frame gives the energy at each frequency bin.
5. Mel filterbank and log. The bins are collapsed into forty mel-spaced triangular filters, and the log
   of each filter's energy is taken, which mirrors the roughly logarithmic loudness response of the ear
   and turns the multiplicative source-filter relationship into an additive one.
6. Discrete cosine transform. The DCT decorrelates the forty log energies and concentrates the
   spectral-envelope shape into the low-order coefficients.

The result for a one-second clip at 16 kHz, using forty mel coefficients and a hop that yields
thirty-two frames, is a forty-by-thirty-two matrix. The vertical axis is spectral shape. The horizontal
axis is time.

### **5.3 - Two Feature Representations: Mean-Pooled and Full Spectrogram**

The forty-by-thirty-two matrix admits two obvious reductions to a feature vector.

**Representation A, mean-pooled.** Average each mel coefficient across all thirty-two frames, giving
forty numbers per clip. This is the reduction most tutorials reach for, and it is defensible: it is
compact, it is invariant to where in the window the word falls, and it captures the average spectral
envelope, which is what distinguishes one vowel from another.

**Representation B, the full flattened spectrogram.** Flatten the matrix into 1,280 numbers per clip,
preserving every frame. This costs thirty-two times the input width and gives up the alignment
invariance that pooling provided.

The two differ in exactly one thing. Representation A retains the average spectrum and destroys its
time-course. Representation B retains both. Any accuracy gap between them, measured under an identical
protocol with identical models, is therefore attributable to the time-course and to nothing else. That
is what makes this pair a controlled experiment rather than a benchmark.

### **5.4 - Why the Split Must Be Speaker-Disjoint**

Speech Commands names every file as `<word>/<speaker_hash>_nohash_<index>.wav`, where the speaker hash
is an eight-character hexadecimal identifier derived from the contributor. A single speaker contributed
many clips across many words.

A random split over clips therefore places recordings of the same voice in both the training and the
test partitions. MFCCs encode the vocal tract as much as they encode the word, so a model given that
split can identify the speaker and use speaker-conditional shortcuts, which inflates the test score
without improving anything a deployed system would care about. The inflation is largest for the
highest-capacity model, which reverses the ranking the experiment is trying to measure.

Section 8.6 partitions on the speaker hash. Every clip from a given speaker lands in exactly one of
train, validation, and test. The resulting numbers are lower than a random split would report and they
are the ones that transfer.

### **7.2 - Downloading and Indexing the Clips**

```python
SPEECH_COMMANDS_URL = (
    "https://storage.googleapis.com/download.tensorflow.org/data/"
    "speech_commands_v0.02.tar.gz"
)

# TODO(numbers): record the extracted clip count and the on-disk size after the
# first successful run.

def build_index(root: pathlib.Path, words: list[str]) -> pd.DataFrame:
    '''Index every clip as (path, word, speaker_hash).

    Speech Commands encodes the contributor in the filename as
    <speaker_hash>_nohash_<index>.wav. The hash is what section 8.6 splits on.
    '''
    pattern = re.compile(r"^([0-9a-f]+)_nohash_(\d+)\.wav$")
    rows = []
    for word in words:
        for wav in sorted((root / word).glob("*.wav")):
            match = pattern.match(wav.name)
            if match is None:
                continue
            rows.append({"path": str(wav), "word": word,
                         "speaker": match.group(1)})
    return pd.DataFrame(rows)
```

### **8.3 - MFCC Extraction**

```python
SAMPLE_RATE = 16_000
N_MFCC = 40
N_FFT = 512
HOP_LENGTH = 512   # 16000 / 512 -> 32 frames for a one-second clip
N_FRAMES = 32


def extract_mfcc(path: str) -> np.ndarray:
    '''Return a (N_MFCC, N_FRAMES) MFCC matrix for one clip.

    Clips shorter than one second are zero-padded on the right so that every
    matrix has the same shape. Longer clips are truncated.
    '''
    signal, _ = librosa.load(path, sr=SAMPLE_RATE, mono=True)
    if len(signal) < SAMPLE_RATE:
        signal = np.pad(signal, (0, SAMPLE_RATE - len(signal)))
    signal = signal[:SAMPLE_RATE]
    mfcc = librosa.feature.mfcc(
        y=signal, sr=SAMPLE_RATE, n_mfcc=N_MFCC,
        n_fft=N_FFT, hop_length=HOP_LENGTH,
    )
    return mfcc[:, :N_FRAMES]
```

### **8.4 - Representation A: Mean-Pooled Coefficients**

```python
def mean_pooled(mfcc: np.ndarray) -> np.ndarray:
    '''Collapse the time axis. Forty numbers per clip.'''
    return mfcc.mean(axis=1)
```

### **8.5 - Representation B: The Full Flattened Spectrogram**

```python
def flattened(mfcc: np.ndarray) -> np.ndarray:
    '''Keep every frame. 40 * 32 = 1,280 numbers per clip.

    Row-major order, so index i * N_FRAMES + t is mel coefficient i at frame t.
    Section 14 relies on that layout to fold importance back onto the grid.
    '''
    return mfcc.reshape(-1)
```

### **8.6 - Speaker-Disjoint Train, Validation, and Test Split**

Warden's own release specifies a reproducible partition rule: hash the speaker identifier and assign
the resulting bucket to training, validation, or testing, so that the partition is stable as the
dataset grows and no speaker crosses a boundary. The implementation below follows that rule at an
80 / 10 / 10 target.

TODO(numbers): record the realized partition sizes in clips and in speakers, and confirm all twenty
words appear in all three partitions.

```python
VAL_PERCENTAGE = 10.0
TEST_PERCENTAGE = 10.0
MAX_PER_CLASS = 2 ** 27 - 1


def which_set(speaker: str) -> str:
    '''Assign a speaker to train, validation, or test, deterministically.

    Hashing the speaker rather than the filename is what keeps a speaker whole.
    '''
    digest = hashlib.sha1(speaker.encode("utf-8")).hexdigest()
    bucket = (int(digest, 16) % (MAX_PER_CLASS + 1)) * (100.0 / MAX_PER_CLASS)
    if bucket < VAL_PERCENTAGE:
        return "validation"
    if bucket < VAL_PERCENTAGE + TEST_PERCENTAGE:
        return "test"
    return "train"
```

## **9 - GenAI-Generated Features**

### **9.1 - Overview**

A free NVIDIA-hosted language model receives the feature definitions from section 8, never the data,

The candidates a speech engineer would expect the model to reach for are delta and delta-delta
coefficients, which Sadaoki Furui introduced in 1986 and which encode the first and second time
derivatives of each mel coefficient, along with frame-level energy, zero-crossing rate, and spectral
centroid statistics.

The delta coefficients are worth calling out in advance. They are an explicit, hand-built encoding of
exactly the time-course information that section 5.3 says the mean-pooled representation destroys. Every
AI-suggested feature here is appended to the mean-pooled representation, not to the full spectrogram,
which turns this stage into a second, independent measurement of the same question section 15 asks. If
adding time-derivative statistics to forty pooled coefficients recovers a meaningful part of the gap to
the full 1,280-feature representation, that is corroboration by a different route. If it recovers
nearly all of the gap, the practical recommendation changes, because deltas cost far less than
thirty-two times the input width.

```python
def build_ai_features(mfcc: np.ndarray, specs) -> np.ndarray:
    '''Evaluate each AI-suggested expression against one clip's MFCC matrix.

    Every result is flattened and concatenated, so the returned vector has a
    fixed width provided the spec list is fixed. A spec that raises, returns a
    non-finite value, or varies in width between clips is dropped by
    validate_ai_specs below before this is used at scale.
    '''
    parts = []
    for spec in specs:
        value = eval(spec["expression"], {"np": np}, {"mfcc": mfcc})
        parts.append(np.atleast_1d(np.asarray(value, dtype=float)).ravel())
    return np.concatenate(parts) if parts else np.zeros(0)


def validate_ai_specs(specs, sample_mfccs):
    '''Keep only specs that evaluate cleanly and at constant width across samples.

    An LLM-proposed expression can reference a shape that happens to work on one
    clip and not another, so each candidate is tried on several clips before it
    is trusted. This runs once, on a handful of clips, not on the full dataset.
    '''
    kept = []
    for spec in specs:
        widths = set()
        ok = True
        for mfcc in sample_mfccs:
            try:
                value = eval(spec["expression"], {"np": np}, {"mfcc": mfcc})
                arr = np.atleast_1d(np.asarray(value, dtype=float)).ravel()
                if not np.all(np.isfinite(arr)):
                    raise ValueError("non-finite value")
                widths.add(arr.shape[0])
            except Exception as exc:
                print("Dropping feature", spec["name"], "due to error:", exc)
                ok = False
                break
        if ok and len(widths) == 1:
            kept.append(spec)
        elif ok:
            print("Dropping feature", spec["name"], "for inconsistent width:", widths)
    return kept


# TODO(numbers): run validate_ai_specs on a sample of clips, report which specs
# survived, and record the resulting AI feature-block width.
```

## **10 - GenAI Synthetic Data Augmentation**

### **10.1 - Overview**

Audio augmentation has a well-established vocabulary, so the language model proposes parameter ranges
and the notebook applies them. The candidate transforms are time shift, time stretch, pitch shift, and

Two constraints govern the design, and both are specific to this case study.

First, augmentation must not cross the speaker boundary. A clip augmented from a training speaker stays
in training. Augmenting a test clip and scoring against it would reintroduce exactly the leakage
section 5.4 exists to prevent, in a form that is harder to notice.

Second, time shift and time stretch both perturb the time axis, which is the axis section 15 is about.
Teaching the model invariance to a shift is useful for deployment and it also flattens part of the
signal the law test measures. The augmented run is therefore reported alongside the unaugmented run
rather than replacing it, and section 15 draws its conclusion from the unaugmented run.

### **10.2 - Sampling Synthetic Clips from the LLM-Proposed Parameters**

```python
AUGMENT_PROMPT = '''You are helping design a synthetic data augmentation plan for a spoken-word
recognition dataset. The clips are one second of 16 kHz speech, recorded by volunteers through browsers
and phones, so they already vary in loudness, background noise, microphone, and where in the one-second
window the word falls.

Propose realistic parameter ranges for four standard audio augmentations. Stay within ranges that keep
the word clearly recognisable to a human listener; the goal is realistic variation, not corruption.

Respond with ONLY a JSON array of exactly 4 objects, no other text, in this exact shape:
{"kind": "time_shift" or "time_stretch" or "pitch_shift" or "background_noise",
 "low": <float>, "high": <float>, "unit": "<seconds|rate|semitones|snr_db>"}
'''

try:
    augment_response = nvidia_chat(AUGMENT_PROMPT, max_tokens=500, temperature=0.3)
except Exception as exc:
    print("NVIDIA call unavailable:", exc)
    augment_response = None

FALLBACK_AUGMENT_PLAN = [
    {"kind": "time_shift", "low": -0.10, "high": 0.10, "unit": "seconds"},
    {"kind": "time_stretch", "low": 0.90, "high": 1.10, "unit": "rate"},
    {"kind": "pitch_shift", "low": -2.0, "high": 2.0, "unit": "semitones"},
    {"kind": "background_noise", "low": 5.0, "high": 20.0, "unit": "snr_db"},
]

VALID_KINDS = {"time_shift", "time_stretch", "pitch_shift", "background_noise"}


def parse_augment_plan(text):
    if text is None:
        return None
    match = re.search(r"\[.*\]", text, re.DOTALL)
    if not match:
        return None
    try:
        parsed = json.loads(match.group(0))
        assert isinstance(parsed, list) and len(parsed) > 0
        for item in parsed:
            assert item.get("kind") in VALID_KINDS
            assert float(item["low"]) <= float(item["high"])
        return parsed
    except Exception:
        return None


augment_plan = parse_augment_plan(augment_response)
if augment_plan is None:
    print("Using fallback augmentation plan (no live NVIDIA response parsed).")
    augment_plan = FALLBACK_AUGMENT_PLAN
else:
    print("Parsed", len(augment_plan), "augmentation ranges from the NVIDIA response.")

for region in augment_plan:
    print(f" - {region['kind']}: {region['low']} to {region['high']} {region['unit']}")
```

```python
def load_waveform(path: str) -> np.ndarray:
    '''Load and pad or truncate one clip to exactly one second.'''
    signal, _ = librosa.load(path, sr=SAMPLE_RATE, mono=True)
    if len(signal) < SAMPLE_RATE:
        signal = np.pad(signal, (0, SAMPLE_RATE - len(signal)))
    return signal[:SAMPLE_RATE]


def apply_augmentation(signal: np.ndarray, spec: dict, rng, noise_bank=None) -> np.ndarray:
    '''Apply one augmentation to a waveform, sampling its magnitude from the LLM-proposed range.

    Returns a one-second waveform in every branch, so the downstream MFCC
    extraction keeps its fixed (40, 32) shape.
    '''
    low, high = float(spec["low"]), float(spec["high"])
    kind = spec["kind"]

    if kind == "time_shift":
        shift = int(rng.uniform(low, high) * SAMPLE_RATE)
        out = np.roll(signal, shift)
        # np.roll wraps, which would move the tail of a word to the front of the
        # clip and create a sound no speaker produced. Zero the wrapped region.
        if shift > 0:
            out[:shift] = 0.0
        elif shift < 0:
            out[shift:] = 0.0
        return out

    if kind == "time_stretch":
        rate = rng.uniform(low, high)
        out = librosa.effects.time_stretch(y=signal, rate=rate)
    elif kind == "pitch_shift":
        steps = rng.uniform(low, high)
        out = librosa.effects.pitch_shift(y=signal, sr=SAMPLE_RATE, n_steps=steps)
    elif kind == "background_noise":
        snr_db = rng.uniform(low, high)
        if noise_bank:
            noise = noise_bank[rng.integers(len(noise_bank))]
            start = rng.integers(0, max(1, len(noise) - SAMPLE_RATE))
            noise = noise[start:start + SAMPLE_RATE]
            if len(noise) < SAMPLE_RATE:
                noise = np.pad(noise, (0, SAMPLE_RATE - len(noise)))
        else:
            noise = rng.normal(0, 1, SAMPLE_RATE)
        sig_power = np.mean(signal ** 2) + 1e-12
        noise_power = np.mean(noise ** 2) + 1e-12
        scale = np.sqrt(sig_power / (noise_power * (10 ** (snr_db / 10.0))))
        out = signal + scale * noise
    else:
        out = signal

    if len(out) < SAMPLE_RATE:
        out = np.pad(out, (0, SAMPLE_RATE - len(out)))
    return out[:SAMPLE_RATE]


# TODO(numbers): generate the augmented training set from TRAINING SPEAKERS ONLY,
# re-extract MFCCs through the section 8 pipeline, and record the augmented count.
```

## **12 - Light Deep Learning Model: PyTorch Feedforward Network**

### **12.1 - Overview**

The deep learning model is a feedforward network, chosen deliberately over a convolutional network. A
CNN would exploit the two-dimensional locality of the spectrogram and would very likely score higher,
and it would also make the representation comparison uninterpretable, because the convolution encodes
the time axis structurally. The feedforward network receives the flattened vector with no architectural
knowledge that neighbouring entries are adjacent in time. Whatever it learns about the time-course, it
learns from the data.

The network is trained on both representations, which gives the four-model comparison two complete rows
rather than one.

```python
class SpokenWordFFNN(nn.Module):
    '''Feedforward classifier over a flat MFCC feature vector.

    Deliberately architecture-agnostic about the time axis. See 12.1.
    '''

    def __init__(self, n_features: int, n_classes: int = 20,
                 hidden: tuple[int, ...] = (512, 256, 128), dropout: float = 0.3):
        super().__init__()
        layers: list[nn.Module] = []
        prev = n_features
        for width in hidden:
            layers += [nn.Linear(prev, width), nn.BatchNorm1d(width),
                       nn.ReLU(), nn.Dropout(dropout)]
            prev = width
        layers.append(nn.Linear(prev, n_classes))
        self.net = nn.Sequential(*layers)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.net(x)


# TODO(numbers): training loop, early stopping on the validation partition,
# and the recorded test scores for both representations.
```

### **13.2 - Extracting Pretrained Speech Embeddings**

```python
# !pip install -q transformers

from transformers import Wav2Vec2Model, Wav2Vec2FeatureExtractor
from sklearn.linear_model import LogisticRegression

W2V_MODEL_NAME = "facebook/wav2vec2-base"
W2V_BATCH_SIZE = 32


def load_wav2vec():
    '''Load the frozen encoder and its matching feature extractor.

    eval() plus no_grad at call time keeps dropout off and skips the backward
    graph. Nothing here is fine-tuned; see 13.1.
    '''
    extractor = Wav2Vec2FeatureExtractor.from_pretrained(W2V_MODEL_NAME)
    model = Wav2Vec2Model.from_pretrained(W2V_MODEL_NAME)
    model.eval()
    for param in model.parameters():
        param.requires_grad = False
    return extractor, model


@torch.no_grad()
def wav2vec_embed(signals, extractor, model, device="cpu"):
    '''Mean-pool the frozen encoder's per-frame hidden states into one vector per clip.

    signals is a list or array of one-second waveforms at SAMPLE_RATE. The
    attention mask is passed through so that padded positions, if any, are
    excluded from the pooling rather than averaged in as zeros.
    '''
    inputs = extractor(list(signals), sampling_rate=SAMPLE_RATE,
                       return_tensors="pt", padding=True)
    inputs = {k: v.to(device) for k, v in inputs.items()}
    hidden = model(**inputs).last_hidden_state          # (batch, frames, 768)

    mask = inputs.get("attention_mask")
    if mask is None:
        return hidden.mean(dim=1).cpu().numpy()

    frame_mask = model._get_feature_vector_attention_mask(
        hidden.shape[1], mask
    ).unsqueeze(-1).to(hidden.dtype)
    pooled = (hidden * frame_mask).sum(dim=1) / frame_mask.sum(dim=1).clamp(min=1e-9)
    return pooled.cpu().numpy()


# TODO(numbers): embed the train and test partitions in batches of W2V_BATCH_SIZE,
# fit LogisticRegression on the training embeddings, and record test accuracy.
# The split stays the speaker-disjoint one from section 8.6.
```

## **14 - Explainability: Permutation Importance over the Spectrogram Grid**

### **14.1 - Why Permutation Importance for This Case**

The explainability method has to answer a question about structure across 1,280 correlated inputs, and
the answer has to fold back onto a forty-by-thirty-two grid. That rules out the defaults used elsewhere
in this repository. SHAP over 1,280 features is expensive and its additive attribution is hard to read
once neighbouring features are as correlated as adjacent spectrogram frames are. LIME, used in CS11,
explains one instance well and does not aggregate cleanly over a grid.

Permutation importance fits. It is model-agnostic, so the same computation runs against XGBoost and the
feedforward network without change, it measures the drop in test accuracy when one input is shuffled,
which is the quantity this section actually wants, and its per-feature scores sum onto the grid axes
without further assumptions. Its known weakness with correlated features is real, and the notebook
handles it by permuting whole rows and whole columns as groups rather than reading individual cells.

### **14.2 - Importance Across Mel Bands**

Summing importance across the time axis for each mel coefficient gives forty numbers, one per band.

The expectation from section 5.1 is that the low-order coefficients dominate, because the DCT
concentrates the spectral envelope there and the first two formants live in that region. A flat profile
across all forty bands would say the model is not using the mel structure at all.

### **14.3 - Importance Across Time Frames**

Summing importance across the mel axis for each frame gives thirty-two numbers, one per time step. This
is the measurement section 15 turns on.

Two outcomes are distinguishable. A flat profile says every frame contributes equally, which is what
would happen if the model were using a time-averaged spectrum reconstructed from the flattened input,
and it would mean the full representation bought nothing structural. A peaked profile says particular
frames carry the decision, and the location of those peaks relative to word onset is then directly
comparable to where a phonetician would point on a spectrogram.

### **15.2 - Mapping the Result to the Haskins Formant-Transition Finding**

In 1955, Pierre Delattre, Alvin Liberman, and Franklin Cooper at Haskins Laboratories hand-painted
spectrograms onto acetate and played them back as sound through the Pattern Playback. By varying one
element at a time, they established something that had not been obvious from listening: the identity of
a stop consonant is carried by the direction and extent of the formant transition into the following
vowel, and the same steady-state spectrum yields a different consonant depending on that transition.
They summarized it as an acoustic locus, a virtual frequency the second formant appears to originate
from, specific to the place of articulation.

The consequence for a feature representation is direct. Average a spectrogram down its time axis and
the transitions cancel. What survives is the average spectral envelope, which distinguishes vowels
reasonably and consonants poorly. Any model restricted to that average is working with the information
Haskins showed is insufficient.

## **16 - Agentic Layer: LangGraph Voice-Command Recommendation**

### **16.1 - Overview**

The agent serves the engineer who owns a deployed wake-word pipeline and has received a report that the
device mishears a particular word. It receives the model's prediction, its confidence, the top
contributing frames and bands from section 14, and the confusion-matrix row for the predicted class,
and it returns a plain-language diagnosis and a concrete recommended action.

The actions worth recommending in this domain are specific. Collecting more speakers helps a word that
is confused across many speakers. Adjusting endpointing, the logic that decides where the one-second
window falls around detected speech, helps a short word whose evidence sits in a few frames that
sometimes fall outside the window. Raising a confidence threshold helps a word whose errors are
confident rather than uncertain. Distinguishing these three from the same evidence is the work.

The backend is a free NVIDIA-hosted language model, following this repository's migration from Hugging
Face after the free tier proved too rate-limited for repeated agentic experimentation. Both the fixed
pipeline and the autonomous variant call it through the shared rate limiter.

```python
from typing import TypedDict, Optional, List
from langgraph.graph import StateGraph, END


class ClipState(TypedDict):
    clip_id: str
    true_word: Optional[str]
    predicted_word: str
    confidence: float
    runner_up_word: str
    runner_up_confidence: float
    top_frames: List[int]
    top_bands: List[int]
    silence_fraction: float
    train_speakers_for_word: int
    diagnosis: Optional[str]
    recommendation: Optional[str]


def diagnose_node(state: ClipState) -> ClipState:
    '''Rule-based diagnosis. Hardcoded branches, no tools, no memory.

    This is deliberately the naive version. Section 16.3 replaces it.
    '''
    margin = state["confidence"] - state["runner_up_confidence"]
    early_frames = [f for f in state["top_frames"] if f <= 4 or f >= N_FRAMES - 5]

    if state["silence_fraction"] > 0.6 and early_frames:
        pattern = ("an alignment problem, where the word sits near the edge of the one-second window "
                   "and the frames carrying it risk falling outside it")
    elif margin < 0.15:
        pattern = (f"a confusable pair, where {state['predicted_word']} and "
                   f"{state['runner_up_word']} compete closely")
    elif state["train_speakers_for_word"] < 100:
        pattern = "thin speaker coverage for this word in the training partition"
    elif state["confidence"] > 0.8:
        pattern = "a confident error, which a confidence threshold alone will not catch"
    else:
        pattern = "a general low-confidence case not clearly tied to one mechanism"

    state["diagnosis"] = (
        f"Clip {state['clip_id']} was predicted as {state['predicted_word']} with confidence "
        f"{state['confidence']:.2f}, ahead of {state['runner_up_word']} at "
        f"{state['runner_up_confidence']:.2f}, a margin of {margin:.2f}. The clip is "
        f"{state['silence_fraction']:.0%} silence. Permutation importance for this prediction "
        f"concentrated on time frames {state['top_frames']} and mel bands {state['top_bands']}. "
        f"The pattern most resembles {pattern}."
    )
    return state


def recommend_node(state: ClipState) -> ClipState:
    prompt = f'''You are a speech recognition engineering assistant. Given this diagnosis of a
misrecognised voice command, write a 3 to 4 sentence recommendation in plain language for the engineer
who owns the on-device wake-word pipeline. Recommend one specific, concrete action appropriate to the
pattern described, for example collecting additional speakers for a confusable word pair, adjusting the
endpointing window so short words are not clipped, or raising the confidence threshold for this class.
Note the confidence implied by the numbers.

Diagnosis:
{state["diagnosis"]}
'''
    try:
        state["recommendation"] = nvidia_chat(prompt, max_tokens=400, temperature=0.2)
    except Exception as exc:
        state["recommendation"] = (
            f"NVIDIA call unavailable ({exc}). The rule-based diagnosis above is the only "
            "output available in this environment."
        )
    return state


fixed_graph = StateGraph(ClipState)
fixed_graph.add_node("diagnose", diagnose_node)
fixed_graph.add_node("recommend", recommend_node)
fixed_graph.set_entry_point("diagnose")
fixed_graph.add_edge("diagnose", "recommend")
fixed_graph.add_edge("recommend", END)
word_agent = fixed_graph.compile()
```

### **16.3 - From Fixed Pipeline to Autonomous Agent**

The pipeline in 16.1 is an agent in name. Its diagnosis is a chain of Python branches a developer wrote
in advance, it calls no tools, it sees only the fields the caller happened to put in the state, and it
remembers nothing between runs. The language model writes prose at the end of a decision that was
already made.

The autonomous variant inverts that. The model receives a goal and a set of tools, and it decides what
to look at. It can check the confusion row before or after pulling the clip's features, it can skip a
tool it judges irrelevant, it can call one twice, and it can consult a log of previous investigations
before reaching a conclusion. The branches move out of Python and into the model's own planning.

The comparison is the point of the stage. Section 18 reports where the autonomous run reached a
different conclusion than the fixed pipeline and whether the extra latitude was worth it, including the
cases where it was not.

### **16.4 - Tools and Memory for the Autonomous Agent**

Six tools, each reading a real measurement rather than a canned string. The memory is an append-only
JSONL file under `data/`, which `.gitignore` excludes, so the log persists across runs of the notebook
without entering version control.

```python
AGENT_MEMORY_PATH = DATA_DIR / "agent_memory.jsonl"

# Runtime registry so the same tools can answer questions about a clip that was
# never part of the test set, for example one supplied by section 17's demo.
RUNTIME_CLIP_REGISTRY = {}


def get_clip_features(clip_id: str) -> dict:
    '''Return the acoustic summary for one clip: duration, silence, and MFCC statistics.'''
    if clip_id in RUNTIME_CLIP_REGISTRY:
        return RUNTIME_CLIP_REGISTRY[clip_id]["features"]
    # TODO(numbers): look the clip up in the test partition and return its
    # measured silence fraction, energy, and per-band MFCC means.
    return {"error": f"Clip {clip_id} not found."}


def predict_word(clip_id: str) -> dict:
    '''Return the trained model's prediction, confidence, and top three candidates.'''
    if clip_id in RUNTIME_CLIP_REGISTRY:
        return RUNTIME_CLIP_REGISTRY[clip_id]["prediction"]
    # TODO(numbers): score the clip with the reference model and return the
    # predicted word, its probability, and the runners-up.
    return {"error": f"Clip {clip_id} not found."}


def confusion_row(word: str) -> dict:
    '''Return which words this word is most often confused with, from the test confusion matrix.'''
    if word not in LABEL_TO_ID:
        return {"error": f"{word} is not one of the twenty words."}
    # TODO(numbers): return the row of the measured confusion matrix.
    return {"word": word, "confused_with": "TODO(numbers)"}


def speaker_count(word: str) -> dict:
    '''Return how many distinct training speakers recorded this word.'''
    if word not in LABEL_TO_ID:
        return {"error": f"{word} is not one of the twenty words."}
    # TODO(numbers): count distinct speakers for this word in the training partition.
    return {"word": word, "train_speakers": "TODO(numbers)"}


def recall_similar_incidents(word: str, limit: int = 3) -> dict:
    '''Read past investigations of this word from the append-only memory log.'''
    if not AGENT_MEMORY_PATH.exists():
        return {"word": word, "past_incidents": []}
    records = []
    with open(AGENT_MEMORY_PATH) as fh:
        for line in fh:
            line = line.strip()
            if not line:
                continue
            try:
                record = json.loads(line)
            except json.JSONDecodeError:
                continue
            if record.get("word") == word:
                records.append(record)
    return {"word": word, "past_incidents": records[-limit:]}


def flag_for_pipeline_owner(word: str, reason: str) -> dict:
    '''Escalate to the pipeline owner and log the session. The agent's final step.'''
    record = {"word": word, "reason": reason, "timestamp": time.time()}
    AGENT_MEMORY_PATH.parent.mkdir(parents=True, exist_ok=True)
    with open(AGENT_MEMORY_PATH, "a") as fh:
        fh.write(json.dumps(record) + "\n")
    return {"status": "escalated", "word": word, "logged_to": str(AGENT_MEMORY_PATH)}


AGENT_TOOL_REGISTRY = {
    "get_clip_features": get_clip_features,
    "predict_word": predict_word,
    "confusion_row": confusion_row,
    "speaker_count": speaker_count,
    "recall_similar_incidents": recall_similar_incidents,
    "flag_for_pipeline_owner": flag_for_pipeline_owner,
}


def _tool_schema(name, description, properties, required):
    return {"type": "function", "function": {
        "name": name, "description": description,
        "parameters": {"type": "object", "properties": properties, "required": required}}}


CLIP_ARG = {"clip_id": {"type": "string", "description": "The clip identifier."}}
WORD_ARG = {"word": {"type": "string", "description": "One of the twenty vocabulary words."}}

AGENT_TOOL_SCHEMAS = [
    _tool_schema("get_clip_features", "Acoustic summary for one clip.", CLIP_ARG, ["clip_id"]),
    _tool_schema("predict_word", "Model prediction and confidence for one clip.", CLIP_ARG, ["clip_id"]),
    _tool_schema("confusion_row", "What this word is confused with.", WORD_ARG, ["word"]),
    _tool_schema("speaker_count", "Distinct training speakers for this word.", WORD_ARG, ["word"]),
    _tool_schema("recall_similar_incidents", "Past investigations of this word.", WORD_ARG, ["word"]),
    _tool_schema(
        "flag_for_pipeline_owner",
        "Escalate to the pipeline owner. Call exactly once, as the final step.",
        {**WORD_ARG, "reason": {"type": "string", "description": "The escalation reasoning."}},
        ["word", "reason"],
    ),
]
```

### **16.5 - Building the ReAct Loop with Conditional Edges**

The loop is two nodes and one conditional edge. The agent node calls the model with the tool schemas
attached. The conditional edge inspects the reply: a reply carrying tool calls routes to the tools node
and then back to the agent, and a reply carrying only prose routes to the end. The recursion limit caps
how many times that cycle can run, which is what stops a model that keeps calling tools forever.

The rate limiter matters more here than anywhere else in the notebook, because a ReAct loop issues
several calls per investigation rather than one.

```python
import requests
from nvidia_rate_limited_call import _acquire_slot

NIM_URL = "https://integrate.api.nvidia.com/v1/chat/completions"
AGENT_MODEL = "nvidia/nemotron-3-ultra-550b-a55b"


def _nvidia_api_key():
    env_path = os.path.expanduser("~/.config/secrets/.env")
    if not os.path.exists(env_path):
        return None
    with open(env_path) as fh:
        for line in fh:
            if line.startswith("NVIDIA_API_KEY="):
                return line.strip().split("=", 1)[1]
    return None


def nvidia_chat_tools(messages, tools, max_tokens=600, temperature=0.2):
    '''Tool-calling counterpart to nvidia_chat, sharing the same rate-limit slot.

    nvidia_chat sends a single prompt and returns text. A ReAct loop needs the
    full message object back, including any tool_calls, so this posts the
    conversation and returns the raw assistant message.
    '''
    api_key = _nvidia_api_key()
    if api_key is None:
        raise RuntimeError("No NVIDIA_API_KEY configured.")
    _acquire_slot()
    resp = requests.post(
        NIM_URL,
        headers={"Authorization": f"Bearer {api_key}", "Content-Type": "application/json"},
        json={"model": AGENT_MODEL, "messages": messages, "tools": tools,
              "tool_choice": "auto", "max_tokens": max_tokens, "temperature": temperature},
        timeout=(30, 150),
    )
    resp.raise_for_status()
    return resp.json()["choices"][0]["message"]


class AgentState(TypedDict):
    messages: List[dict]


def agent_node(state: AgentState) -> AgentState:
    try:
        state["messages"].append(nvidia_chat_tools(state["messages"], AGENT_TOOL_SCHEMAS))
    except Exception as exc:
        state["messages"].append({
            "role": "assistant",
            "content": f"Tool-calling request failed ({exc}). Stopping the autonomous loop.",
        })
    return state


def tools_node(state: AgentState) -> AgentState:
    for tool_call in state["messages"][-1].get("tool_calls") or []:
        fn_name = tool_call["function"]["name"]
        try:
            fn_args = json.loads(tool_call["function"]["arguments"])
        except json.JSONDecodeError:
            fn_args = {}
        fn = AGENT_TOOL_REGISTRY.get(fn_name)
        if fn is None:
            result = {"error": f"Unknown tool {fn_name}"}
        else:
            try:
                result = fn(**fn_args)
            except Exception as exc:
                result = {"error": str(exc)}
        state["messages"].append({
            "role": "tool",
            "tool_call_id": tool_call.get("id", fn_name),
            "name": fn_name,
            "content": json.dumps(result),
        })
    return state


def route_after_agent(state: AgentState) -> str:
    return "tools" if state["messages"][-1].get("tool_calls") else "end"


autonomous_graph = StateGraph(AgentState)
autonomous_graph.add_node("agent", agent_node)
autonomous_graph.add_node("tools", tools_node)
autonomous_graph.set_entry_point("agent")
autonomous_graph.add_conditional_edges("agent", route_after_agent,
                                       {"tools": "tools", "end": END})
autonomous_graph.add_edge("tools", "agent")
autonomous_word_agent = autonomous_graph.compile()
```
