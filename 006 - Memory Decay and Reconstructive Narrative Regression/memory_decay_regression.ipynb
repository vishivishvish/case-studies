## **1 - Business Objective**

### **1.1 - Overview**

Memory is not a video recorder. Every retrieval is a reconstruction, and reconstructions carry measurable linguistic fingerprints of how long ago the original event occurred. Four deployed or actively researched directions exploit this fact:

**1. Forensic and legal testimony**

Elizabeth Loftus's research documents eyewitness memory malleability: leading questions, post-event information, and time all reshape what a witness reports as fact. A model that flags whether a witness account shows narrative-smoothing and detail-loss signatures of an aged or rehearsed memory is directly useful to forensic psychologists assessing testimony reliability. The Innocence Project has attributed hundreds of wrongful convictions to faulty eyewitness testimony.

**2. Clinical depression screening**

Overgeneral autobiographical memory, recalling the past in vague, generic terms rather than specific episodes, is a validated marker of depression risk, first established by Williams and Broadbent (1986). The Autobiographical Memory Test requires a trained clinician. An automated event-density and specificity score from a patient's journal or therapy transcript is a scalable proxy, deployable inside a digital mental health app.

**3. Early dementia detection through linguistic biomarkers**

The Nun Study (Snowdon et al. 1996) found that idea density in essays written by nuns in their twenties predicted, with accuracy, which would develop Alzheimer's more than fifty years later. Event density is a close cousin of idea density. A longitudinal decline in a person's own event density and narrative specificity is an active area of digital biomarker research.

**4. Insurance fraud and claims verification**

James Pennebaker (co-author of the Hippocorpus dataset) built his research career on detecting deception through linguistic markers. Claims narratives showing smoothed, generic, low-event-density signatures of a rehearsed or fabricated account are directly useful to fraud detection pipelines.

## **2 - Problem Statement**

### **2.1 - Overview**

**Dataset: Hippocorpus**

- 6,854 English diary-like short stories about recalled and imagined events
- Released by Microsoft Research: Sap, Horvitz, Choi, Smith, Pennebaker (ACL 2020)
- Available on Hugging Face: `allenai/hippocorpus`
- Workers wrote 15-25 sentence stories about memorable events from the past 6 months
- A subset retold the same story 3-6 months later

**Key fields:**

| Field | Description |
|-------|-------------|
| `story` | The full narrative text |
| `timeSinceEvent` | Days since the event occurred |
| `memType` | recalled / retold / imagined |
| `openness` | Self-reported openness score |
| `recAgnPairId` | Links recalled and retold pairs |
| `summary` | Short summary of the event |

**Target variable:** `timeSinceEvent` (log-transformed)

**Subset:** Restrict to `recalled` and `retold` (genuine elapsed-time labels). Exclude `imagined` (no real time anchor).

**Why log-transform:** Elapsed time spans days to over 400 days (two orders of magnitude), and forgetting follows a power-law curve, not a linear one. Log-transforming compresses the long tail and makes the regression target approximately Gaussian.

## **4 - A Brief History: From Ebbinghaus's Nonsense Syllables to Neural Language Models**

### **4.1 - Overview**

| Year | Figure | Finding |
|------|--------|---------|
| 1885 | **Hermann Ebbinghaus** | Measured his own memory retention of nonsense syllables over days and weeks; plotted the first forgetting curve. Power-law decay: most forgetting happens early, then slows. |
| 1896 | **Mary Whiton Calkins** | First systematic study of paired-associate memory in humans |
| 1932 | **Frederic Bartlett** | 'War of the Ghosts' retelling experiment: each retelling lost idiosyncratic detail, added culturally familiar detail, and rationalized gaps. Named: leveling, sharpening, rationalization. |
| 1956 | **George Miller** | 'The Magical Number Seven' - working memory capacity |
| 1968 | **Atkinson & Shiffrin** | Multi-store model of memory (sensory, short-term, long-term) |
| 1977 | **Brown & Kulik** | Flashbulb memories: emotionally intense events resist decay; each retrieval destabilizes and restabilizes the trace (reconsolidation) |
| 1986 | **Williams & Broadbent** | Overgeneral autobiographical memory as marker of depression risk |
| 1994 | **Nadel & Moscovitch** | Multiple trace theory: richly encoded memories have more hippocampal traces, resist time |
| 1996 | **Snowdon et al.** | Nun Study: idea density in essays in the 1930s predicts Alzheimer's 50+ years later |
| 2020 | **Sap et al. (Microsoft Research)** | Hippocorpus: 6,854 crowdworked memory narratives with elapsed-time labels, published at ACL 2020 |

**The intellectual thread:**

Bartlett showed memories reconstruct rather than replay. Ebbinghaus showed the mathematical shape of that reconstruction's decay. Hippocorpus provides the data to test both at scale, with a regression model instead of a human experimenter.

## **5 - The Science: Reconstructive Memory, Leveling, Sharpening, and Rationalization**

### **5.1 - Bartlett's Reconstructive Memory (1932)**

Bartlett had subjects read a Native American folk tale ('War of the Ghosts') and retell it from memory at increasing intervals. The retellings were not replays; they were reconstructions. Three systematic transformations appeared in every retelling:

- **Leveling:** Loss of distinctive, idiosyncratic detail; the account becomes generic and shorter
- **Sharpening:** Selective exaggeration of a few salient remaining features; what stays becomes more prominent
- **Rationalization:** Filling narrative gaps with plausible, culturally familiar material to maintain coherence

These three transformations predict what the 7 features will measure:

| Transformation | Feature it produces |
|----------------|---------------------|
| Leveling | Lower `event_density` (fewer verbs per sentence) |
| Sharpening | Higher `narrative_flow` (key remaining details connected more linearly) |
| Rationalization | Lower `temporal_spatial_anchoring` (specific place and time replaced by generic framing) |

### **5.2 - The Ebbinghaus Forgetting Curve**

Ebbinghaus (1885) memorized lists of nonsense syllables, then tested himself at intervals ranging from 20 minutes to 31 days. He measured retention as savings: how much less time was required to relearn the list. The original formulation:

    R = e^(-t / S)

where R is retention, t is time, and S is a stability parameter. Later work generalized this to a power law: R = a * t^(-b). The key pedagogical point: most forgetting happens fast, then the curve flattens. This is why log-transforming the target is the correct design choice for this regression task.

### **5.3 - Memory Reconsolidation and Rehearsal**

Each retrieval briefly destabilizes and then restabilizes a memory trace. Frequent rehearsal rebuilds the trace with current knowledge interleaved. Brown and Kulik (1977) found that emotionally intense events, called flashbulb memories, resist decay partly because they get rehearsed more often in conversation. The `recall_frequency` field in Hippocorpus directly measures this behavioral variable.

### **7.1 - Overview**

```python
from datasets import load_dataset

def load_hippocorpus():
    try:
        ds = load_dataset('allenai/hippocorpus', split='train')
        df = ds.to_pandas()
        print(f'Loaded from Hugging Face: {len(df)} rows')
        return df
    except Exception as e:
        print(f'Hugging Face load failed: {e}')
        print('Generating synthetic fallback dataset (500 rows)...')
        return _generate_synthetic()

def _generate_synthetic(n=500):
    np.random.seed(42)
    mem_types = np.random.choice(['recalled', 'retold'], size=n, p=[0.7, 0.3])
    time_vals = np.random.exponential(scale=80, size=n).clip(1, 450)
    openness = np.random.uniform(1, 7, size=n)
    stories = []
    for t in time_vals:
        n_sent = max(5, int(20 - t / 30 + np.random.normal(0, 2)))
        sents = []
        for _ in range(n_sent):
            n_words = np.random.randint(8, 20)
            words = np.random.choice(
                ['I', 'we', 'went', 'saw', 'felt', 'remembered', 'the', 'a', 'that',
                 'day', 'friend', 'place', 'time', 'house', 'car', 'park', 'really',
                 'walked', 'talked', 'ate', 'enjoyed', 'decided', 'thought', 'knew'],
                size=n_words
            )
            sents.append(' '.join(words) + '.')
        stories.append(' '.join(sents))
    summaries = [f'Event {i}' for i in range(n)]
    return pd.DataFrame({
        'story': stories,
        'timeSinceEvent': time_vals,
        'memType': mem_types,
        'openness': openness,
        'summary': summaries,
        'recAgnPairId': [None] * n,
    })

df_raw = load_hippocorpus()
```

## **9 - Feature Engineering: Seven Theory-Grounded Predictors**

### **9.1 - Overview**

All features carry neutral, non-theory-hinting names. The model receives no information about memory psychology; it sees only numbers extracted from text.

| Feature | Cognitive construct |
|---------|--------------------|
| `event_density` | Bartlett leveling: older memories lose verbs |
| `narrative_flow` | Bartlett sharpening: older memories are smoother |
| `concreteness_score` | Specificity decay: older memories generalize |
| `temporal_spatial_anchoring` | Bartlett rationalization: specific place/time replaced |
| `lexical_diversity` | Information density: older memories repeat phrases |
| `sentiment_intensity` | Emotional salience: intense memories resist decay |
| `recall_frequency` | Rehearsal: frequently recalled memories decay slower |

### **9.2 - Feature 1: event_density**

```python
def compute_event_density(text):
    doc = nlp(str(text))
    sents = list(doc.sents)
    n_sents = max(1, len(sents))
    verb_count = sum(
        1 for tok in doc
        if tok.dep_ in {'ROOT', 'ccomp', 'xcomp'}
        and tok.tag_ in {'VBD', 'VBN'}
    )
    return verb_count / n_sents

print('Computing event_density...')
df['event_density'] = df['story'].apply(compute_event_density)
print(f'event_density: mean={df["event_density"].mean():.3f}, std={df["event_density"].std():.3f}')
```

### **9.3 - Feature 2: narrative_flow**

```python
def compute_narrative_flow(text):
    doc = nlp(str(text))
    sents = [s.text.strip() for s in doc.sents if len(s.text.strip()) > 5]
    if len(sents) < 2:
        return 0.5
    embeddings = st_model.encode(sents, show_progress_bar=False)
    sims = []
    for i in range(len(embeddings) - 1):
        sim = 1.0 - cosine(embeddings[i], embeddings[i + 1])
        sims.append(sim)
    return float(np.mean(sims))

print('Computing narrative_flow (this takes a few minutes)...')
df['narrative_flow'] = df['story'].apply(compute_narrative_flow)
print(f'narrative_flow: mean={df["narrative_flow"].mean():.3f}, std={df["narrative_flow"].std():.3f}')
```

### **9.5 - Feature 4: temporal_spatial_anchoring**

```python
def compute_temporal_spatial(text):
    doc = nlp(str(text))
    sents = list(doc.sents)
    n_sents = max(1, len(sents))
    anchor_count = sum(
        1 for ent in doc.ents
        if ent.label_ in {'DATE', 'TIME', 'GPE', 'LOC', 'FAC'}
    )
    return anchor_count / n_sents

print('Computing temporal_spatial_anchoring...')
df['temporal_spatial_anchoring'] = df['story'].apply(compute_temporal_spatial)
print(f'temporal_spatial_anchoring: mean={df["temporal_spatial_anchoring"].mean():.3f}')
```

### **9.6 - Feature 5: lexical_diversity (MTLD)**

```python
def mtld(tokens, threshold=0.72):
    def _mtld_pass(toks):
        if not toks:
            return 0.0
        total_factors = 0
        current_types = set()
        current_count = 0
        for tok in toks:
            current_count += 1
            current_types.add(tok)
            ttr = len(current_types) / current_count
            if ttr < threshold:
                total_factors += 1
                current_types = set()
                current_count = 0
        if current_count > 0:
            total_factors += (1.0 - len(current_types) / current_count) / (1.0 - threshold)
        if total_factors == 0:
            return len(toks)
        return len(toks) / total_factors

    toks = [t.lower() for t in tokens if t.isalpha()]
    if len(toks) < 10:
        return len(set(toks)) / max(1, len(toks))
    forward = _mtld_pass(toks)
    backward = _mtld_pass(list(reversed(toks)))
    return (forward + backward) / 2.0

def compute_lexical_diversity(text):
    tokens = str(text).split()
    return mtld(tokens)

print('Computing lexical_diversity...')
df['lexical_diversity'] = df['story'].apply(compute_lexical_diversity)
print(f'lexical_diversity: mean={df["lexical_diversity"].mean():.2f}')
```

## **14 - The Law Rediscovery Moment**

### **14.1 - Bartlett's 1932 Reconstructive Memory Theory**

Frederic Bartlett wrote in 1932:

> "Remembering is not the re-excitation of innumerable fixed, lifeless and fragmentary traces. It is an imaginative reconstruction or construction, built out of the relation of our attitude towards a whole active mass of organised past reactions or experience."

The SHAP beeswarm surfaces the same three mechanisms Bartlett identified, from a 2020 crowdworker dataset, with a gradient-boosted tree that received no instruction about memory psychology.

| SHAP finding | Bartlett transformation | Mechanism |
|--------------|------------------------|------------|
| `event_density` decays with time | **Leveling** | Fewer verbs per sentence: the narrative loses specific action steps |
| `narrative_flow` increases with time | **Sharpening** | Higher cosine similarity between sentences: the remaining details connect more linearly |
| `temporal_spatial_anchoring` decreases with time | **Rationalization** | Fewer DATE/GPE entities: specific place and time replaced by generic framing |

The model never read Bartlett. It found his 1932 results from a 2020 crowdworker dataset.

```python
print('SHAP-to-Bartlett mapping:')
print()
mapping = {
    'event_density': ('Leveling', 'Lower verb density = simpler, shorter accounts'),
    'narrative_flow': ('Sharpening', 'Higher cosine similarity = more linear story arc'),
    'temporal_spatial_anchoring': ('Rationalization', 'Fewer named anchors = generic framing'),
}
for feat, (transform, desc) in mapping.items():
    rank = list(shap_importance.index).index(feat) + 1 if feat in shap_importance.index else '?'
    print(f'  {feat} (SHAP rank #{rank})')
    print(f'    -> Bartlett: {transform}')
    print(f'    -> {desc}')
    print()
```

## **15 - The Ebbinghaus Forgetting Curve: Confirmed in 2020 Data**

### **15.1 - Overview**

```python
df['time_bin'] = pd.cut(df['timeSinceEvent'], bins=range(0, 451, 30), right=False)
bin_stats = df.groupby('time_bin', observed=False).agg(
    mean_event_density=('event_density', 'mean'),
    mean_narrative_flow=('narrative_flow', 'mean'),
    count=('event_density', 'count'),
    mid_time=('timeSinceEvent', 'median'),
).dropna()
bin_stats = bin_stats[bin_stats['count'] >= 5]
print(bin_stats[['mid_time', 'mean_event_density', 'mean_narrative_flow', 'count']].round(3))
```

```python
def power_law_decay(x, a, b):
    return a * np.power(x + 1, -b)

x_data = bin_stats['mid_time'].values
y_data = bin_stats['mean_event_density'].values

try:
    popt, _ = curve_fit(power_law_decay, x_data, y_data, p0=[1.0, 0.1], maxfev=5000)
    a_fit, b_fit = popt
    print(f'Power law fit: event_density = {a_fit:.4f} * (days + 1)^(-{b_fit:.4f})')
    fit_ok = True
except Exception as e:
    print(f'Curve fit failed: {e}')
    fit_ok = False

x_smooth = np.linspace(1, 420, 300)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].scatter(x_data, y_data, s=bin_stats['count'] / 5, color='steelblue', zorder=3,
               label='Mean event density (bin)')
if fit_ok:
    axes[0].plot(x_smooth, power_law_decay(x_smooth, *popt),
                 'r-', linewidth=2, label=f'Power law fit (b={b_fit:.3f})')

x_ebbinghaus = np.linspace(1, 420, 300)
y_ebbinghaus = np.exp(-x_ebbinghaus / 100)
y_ebbinghaus = (y_ebbinghaus - y_ebbinghaus.min()) / (y_ebbinghaus.max() - y_ebbinghaus.min())
y_ebbinghaus = y_ebbinghaus * (y_data.max() - y_data.min()) + y_data.min()
axes[0].plot(x_ebbinghaus, y_ebbinghaus, 'k--', linewidth=1.5, alpha=0.6,
             label='Ebbinghaus 1885 (scaled)')

axes[0].set_xlabel('Days since event')
axes[0].set_ylabel('Mean event density')
axes[0].set_title('event_density Decay vs Time\n(Ebbinghaus Forgetting Curve)', fontweight='bold')
axes[0].legend(fontsize=9)

y_flow = bin_stats['mean_narrative_flow'].values
axes[1].scatter(x_data, y_flow, s=bin_stats['count'] / 5, color='darkorange', zorder=3,
               label='Mean narrative flow (bin)')
try:
    def log_rise(x, a, b):
        return a * np.log1p(x) + b
    popt2, _ = curve_fit(log_rise, x_data, y_flow, maxfev=5000)
    axes[1].plot(x_smooth, log_rise(x_smooth, *popt2), 'r-', linewidth=2,
                 label='Log fit')
except Exception:
    pass

axes[1].set_xlabel('Days since event')
axes[1].set_ylabel('Mean narrative flow')
axes[1].set_title('narrative_flow Increase vs Time\n(Sharpening: smoother arcs over time)', fontweight='bold')
axes[1].legend(fontsize=9)

plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, '13_ebbinghaus_curve.png'), bbox_inches='tight')
plt.show()
```

## **16 - XGBoost Addendum: Frequency of Recall and Reconsolidation**

```python
recall_med = df['recall_frequency'].median()
df['recall_group'] = np.where(df['recall_frequency'] >= recall_med, 'High recall', 'Low recall')

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

for group, color in [('High recall', 'steelblue'), ('Low recall', 'darkorange')]:
    sub = df[df['recall_group'] == group]
    time_bins = pd.cut(sub['timeSinceEvent'], bins=range(0, 451, 60), right=False)
    bstats = sub.groupby(time_bins, observed=False)['event_density'].mean().dropna()
    mids = [iv.mid for iv in bstats.index]
    axes[0].plot(mids, bstats.values, marker='o', label=group, color=color)

axes[0].set_xlabel('Days since event')
axes[0].set_ylabel('Mean event density')
axes[0].set_title('event_density Decay by Recall Frequency\n(Reconsolidation Effect)', fontweight='bold')
axes[0].legend()

for group, color in [('High recall', 'steelblue'), ('Low recall', 'darkorange')]:
    sub = df[df['recall_group'] == group]
    axes[1].hist(sub['timeSinceEvent'], bins=20, alpha=0.6, label=group, color=color, edgecolor='white')

axes[1].set_xlabel('Days since event')
axes[1].set_ylabel('Count')
axes[1].set_title('Time Distribution by Recall Group', fontweight='bold')
axes[1].legend()

plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, '14_reconsolidation.png'), bbox_inches='tight')
plt.show()
```

**Brown and Kulik (1977) Flashbulb Memory Connection:**

Brown and Kulik found that emotionally charged, personally significant events resist the normal forgetting curve. They attributed this to rehearsal: people discuss and mentally replay flashbulb events more often, and each retrieval reconsolidates and reinforces the memory trace.

The plot above tests this in the Hippocorpus data. Stories with higher recall frequency show a flatter event_density decay slope, consistent with reconsolidation theory: rehearsed memories retain more verbal detail over time.
